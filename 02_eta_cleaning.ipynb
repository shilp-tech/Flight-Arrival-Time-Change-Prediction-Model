```python
# Import required libraries
import pandas as pd
import os

DATA_FOLDER = "/Users/shilppatel/Desktop/AA_FINAL_PROJECT"

# Load the raw ETA flight operations dataset
df = pd.read_csv(os.path.join(DATA_FOLDER, "ETA_CHANGE_PREDICTION_backup.csv"), low_memory=False)

print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

```python
# Drop columns that are not useful for modeling
# Keeping GMT columns, POST_NBR, MINS_TO_SCHD_DEP_QTY, and CANCEL_IND intentionally
cols_to_drop = [
    # Audit and personnel columns
    'PRSNEL_ID', 'FOS_AGENT_SINE_CD', 'FOS_AGENT_SINE_HOME_CITY_CD',
    'USER_AUDIT_TMS', 'ACTN_CD', 'LNIATA_TXT',
    # System timestamps
    'SF_LOAD_TMS', 'DFW_POST_TMS', 'EFF_GMT_TMS', 'EXP_GMT_TMS',
    # Redundant time columns
    'SCHD_FIRST_LEG_DEP_LCL_DT', 'SCHD_LEG_DEP_LCL_DT', 'SCHD_LEG_DEP_LCL_TM',
    'SCHD_LEG_ARVL_LCL_DT', 'SCHD_LEG_ARVL_LCL_TM',
    # Data leakage - only known after landing
    'LEG_BLOCK_VARNCE_MIN_QTY', 'ACTL_LEG_ARVL_LCL_TMS',
    'ACTL_LEG_ARVL_LCL_DT', 'ACTL_LEG_ARVL_LCL_TM', 'ACTL_LEG_ARVL_GMT_TMS',
    # Data leakage - only known after flight completes
    'TTL_POST_QTY', 'POST_FINAL_IND',
    # Data leakage - only known after diversion happens
    'DIVERT_IND',
    # Not useful for modeling
    'RS_IND', 'FLIGHT_TYPE_CD', 'MULTI_OPS_IND', 'ACTL_AIRCFT_TAIL_CD',
    'SCHD_SUBFLEET_CD', 'ACTL_SUBFLEET_CD', 'ACTL_LEG_DEP_GATE_CD',
    'ACTL_LEG_ARVL_GATE_CD', 'DEP_AIRPRT_ENROUT_REGION_CD',
    'ARVL_AIRPRT_ENROUT_REGION_CD', 'PREV_ETD_LCL_TMS', 'ETD_LCL_TMS',
    'ORIGIN_POST_TMS', 'ACTL_LEG_DEP_GMT_TMS',
]

df.drop(columns=cols_to_drop, inplace=True)
print(f"Columns remaining: {df.shape[1]}")
print(df.columns.tolist())
```

```python
# Remove cancelled flights - they have no arrival outcome to predict
df = df[df['CANCEL_IND'] != 1.0]

# Keep only flights where actual airports match scheduled airports
# This removes diverted flights where the plane landed at a different airport
df = df[
    (df['SCHD_LEG_DEP_AIRPRT_IATA_CD'] == df['ACTL_LEG_DEP_AIRPRT_IATA_CD']) &
    (df['SCHD_LEG_ARVL_AIRPRT_IATA_CD'] == df['ACTL_LEG_ARVL_AIRPRT_IATA_CD'])
]

# Drop columns no longer needed after filtering
df.drop(columns=[
    'CANCEL_IND',
    'ACTL_LEG_DEP_AIRPRT_IATA_CD',
    'ACTL_LEG_ARVL_AIRPRT_IATA_CD',
    'FLIGHT_HUB_EVENT_TYPE_CD',
    'ACTL_LEG_DEP_LCL_DT',
    'ACTL_LEG_DEP_LCL_TM',
], inplace=True)

df.reset_index(drop=True, inplace=True)
print(f"Rows remaining: {df.shape[0]:,}")
print(f"Columns remaining: {df.shape[1]}")
print(df.columns.tolist())
```

```python
# Keep only valid 2-letter airline codes and remove invalid entries
df = df[df['OPERAT_AIRLN_IATA_CD'].str.match(r'^[A-Z]{2}$', na=False)]

# Remove YX airline code - not an AA regional partner
df = df[~df['OPERAT_AIRLN_IATA_CD'].isin(['YX'])]

# Keep only valid 4-digit flight numbers
df['OPERAT_FLIGHT_NBR'] = pd.to_numeric(df['OPERAT_FLIGHT_NBR'], errors='coerce')
df = df[df['OPERAT_FLIGHT_NBR'].between(1000, 9999)]
df['OPERAT_FLIGHT_NBR'] = df['OPERAT_FLIGHT_NBR'].astype(int)

df.reset_index(drop=True, inplace=True)
print(f"Rows remaining: {df.shape[0]:,}")
print(df['OPERAT_AIRLN_IATA_CD'].value_counts())
```

```python
# Convert timestamp columns to datetime
df['SCHD_LEG_DEP_LCL_TMS'] = pd.to_datetime(df['SCHD_LEG_DEP_LCL_TMS'])
df['SCHD_LEG_ARVL_LCL_TMS'] = pd.to_datetime(df['SCHD_LEG_ARVL_LCL_TMS'])

# Calculate scheduled flight duration in minutes
df['flight_duration_min'] = (
    df['SCHD_LEG_ARVL_LCL_TMS'] - df['SCHD_LEG_DEP_LCL_TMS']
).dt.total_seconds() / 60

# Remove flights with impossible durations
df = df[df['flight_duration_min'].between(1, 1440)]
df.reset_index(drop=True, inplace=True)

print(f"Rows remaining: {df.shape[0]:,}")
print(f"Duration stats:")
print(df['flight_duration_min'].describe())
```

```python
# Remove rows where scheduled departure timestamp is missing
df.dropna(subset=['SCHD_LEG_DEP_LCL_TMS'], inplace=True)
df.reset_index(drop=True, inplace=True)

print(f"Rows remaining: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

```python
# Save the cleaned ETA dataset
output_path = os.path.join(DATA_FOLDER, "ETA_CHANGE_PREDICTION_clean.csv")
df.to_csv(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
print(f"Columns: {df.columns.tolist()}")
```
